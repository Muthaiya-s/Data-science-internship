```python
import matplotlib.pyplot as plt

# 1. Set a small canvas size (5 inches wide by 4 inches tall)
plt.figure(figsize=(5, 4))

# 2. Create a horizontal violin plot for G1
# vert=False makes it horizontal, showmedians=True adds a line for the median
vp = plt.violinplot(df['G1'], vert=False, showmedians=True, showmeans=True)

# 3. Customize the colors to match your previous graphs
for body in vp['bodies']:
    body.set_facecolor('#66b3ff')  # Light blue fill
    body.set_edgecolor('black')     # Black outline
    body.set_alpha(0.7)

vp['cmaxes'].set_color('black')     # Max value line
vp['cmins'].set_color('black')      # Min value line
vp['cbars'].set_color('black')      # Center bar line
vp['cmedians'].set_color('red')     # Red line for median
vp['cmeans'].set_color('blue')      # Blue line for mean

# 4. Add titles and labels with matching small font sizes
plt.title('Distribution of G1 Grades (Violin)', fontsize=10)
plt.xlabel('G1 Score', fontsize=8)

# 5. Clean up the axes and add grid lines
plt.yticks([])  # Removes the blank '1' on the y-axis since there's only one plot
plt.xticks(fontsize=8)
plt.grid(True, linestyle='--', alpha=0.5)

# 6. Prevent edge clipping
plt.tight_layout()

# Display the plot
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1. Set up a clean, compact canvas size
fig, ax1 = plt.subplots(figsize=(7, 4))

# 2. Plot G1 and G2 on the left Y-axis (Scale: 0-20)
ax1.scatter(df['age'], df['G1'], color='#66b3ff', s=30, alpha=0.6, edgecolor='k', linewidth=0.5, label='G1 Grade')
ax1.scatter(df['age'], df['G2'], color='#ff9999', s=30, alpha=0.6, edgecolor='k', linewidth=0.5, label='G2 Grade')

ax1.set_xlabel('Age', fontsize=8)
ax1.set_ylabel('Grades (0-20)', fontsize=8)
ax1.set_ylim(-1, 21)
ax1.tick_params(labelsize=8)

# 3. Create a second Y-axis on the right for Health Status (Scale: 1-5)
ax2 = ax1.twinx()
ax2.scatter(df['age'], df['health_status'], color='#99ff99', s=30, alpha=0.6, edgecolor='k', linewidth=0.5, label='Health Status')

ax2.set_ylabel('Health Status (1-5)', fontsize=8)
ax2.set_ylim(0.5, 5.5)
ax2.tick_params(labelsize=8)

# 4. Add titles, grid, and combine legends
plt.title('Health Status, G1, and G2 Grades by Age', fontsize=10, fontweight='bold')

# Gather legends from both axes so they display together
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper right', fontsize=8)

ax1.grid(True, linestyle='--', alpha=0.3)
plt.tight_layout()

# Display the combined scatter plot
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1. Set up a clean, compact canvas size
plt.figure(figsize=(7, 4))

# 2. Create the combined box plot
# We pass the columns as a list, and use labels to name each box on the X-axis
plt.boxplot([df['G1'], df['G2'], df['health_status']],
            labels=['G1 Grade', 'G2 Grade', 'Health Status'],
            patch_artist=True,  # Allows us to fill the boxes with color
            boxprops=dict(facecolor='#66b3ff', color='black', alpha=0.7), # Customize box
            medianprops=dict(color='red', linewidth=1.5),                 # Red line for median
            flierprops=dict(marker='o', markerfacecolor='#ff9999', markersize=5)) # Outliers

# 3. Add title and labels with matching small fonts
plt.title('Distribution Comparison: G1, G2, and Health Status', fontsize=10, fontweight='bold')
plt.ylabel('Scores / Values Range', fontsize=8)

# 4. Clean up the ticks and grid
plt.xticks(fontsize=8)
plt.yticks(range(0, 21, 2), fontsize=8) # Shows ticks from 0 to 20 grouped by 2
plt.grid(True, linestyle='--', alpha=0.3, axis='y') # Horizontal grid lines only
plt.tight_layout()

# Display the combined box plot
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1. Set up a clean, compact canvas size
plt.figure(figsize=(7, 4))

# 2. Create the combined violin plot
# We pass the columns as a list. showmedians=True adds a line for the median.
vp = plt.violinplot([df['G1'], df['G2'], df['health_status']],
                    showmedians=True,
                    showmeans=False)

# 3. Customize colors for each violin body
colors = ['#66b3ff', '#ff9999', '#99ff99']
for body, color in zip(vp['bodies'], colors):
    body.set_facecolor(color)    # Fill color
    body.set_edgecolor('black')  # Outline color
    body.set_alpha(0.7)          # Transparency

# Customize the lines (center bar, max/min, and medians)
vp['cmaxes'].set_color('black')
vp['cmins'].set_color('black')
vp['cbars'].set_color('black')
vp['cmedians'].set_color('red')  # Distinct red line for the medians

# 4. Add title and labels
plt.title('Distribution Comparison: G1, G2, and Health Status (Violin)', fontsize=10, fontweight='bold')
plt.ylabel('Scores / Values Range', fontsize=8)

# 5. Manually set the labels for each violin on the X-axis
# By default, violinplot places them at positions 1, 2, 3...
plt.xticks([1, 2, 3], ['G1 Grade', 'G2 Grade', 'Health Status'], fontsize=8)
plt.yticks(range(0, 21, 2), fontsize=8)

# 6. Add a horizontal grid and clean up layout
plt.grid(True, linestyle='--', alpha=0.3, axis='y')
plt.tight_layout()

# Display the combined violin plot
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1. Calculate the overall average for each column
avg_values = [df['G1'].mean(), df['G2'].mean(), df['health_status'].mean()]
labels = ['Avg G1 Grade', 'Avg G2 Grade', 'Avg Health Status']

# 2. Set up a clean, compact canvas size
plt.figure(figsize=(5, 4))

# 3. Create the pie chart
# autopct='%1.1f%%' formats the slices to show 1 decimal point percentage
plt.pie(avg_values,
        labels=labels,
        autopct='%1.1f%%',
        startangle=90,
        colors=['#66b3ff', '#ff9999', '#99ff99'],
        wedgeprops={'edgecolor': 'black', 'linewidth': 1, 'alpha': 0.8})

# 4. Add a title with matching small fonts
plt.title('Overall Average Comparison', fontsize=10, fontweight='bold')
plt.tight_layout()

# Display the pie chart
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1. Calculate the mean (height) and standard deviation (error) for each column
means = [df['G1'].mean(), df['G2'].mean(), df['health_status'].mean()]
errors = [df['G1'].std(), df['G2'].std(), df['health_status'].std()]
labels = ['G1 Grade', 'G2 Grade', 'Health Status']

# 2. Set up a clean, compact canvas size
plt.figure(figsize=(7, 4))

# 3. Create the bar chart with error bars
# yerr handles the error limits, capsize adds the horizontal lines at the top/bottom of the error bars
bars = plt.bar(labels, means, yerr=errors,
               color=['#66b3ff', '#ff9999', '#99ff99'],
               edgecolor='black',
               alpha=0.8,
               capsize=6,       # Width of the horizontal error caps
               ecolor='red',     # Color of the error bars
               linewidth=1)

# 4. Add title and labels with matching small fonts
plt.title('Average Values with Standard Deviation (Error Bars)', fontsize=10, fontweight='bold')
plt.ylabel('Average Scores / Values', fontsize=8)

# 5. Clean up the ticks and grid
plt.xticks(fontsize=8)
plt.yticks(range(0, 21, 2), fontsize=8) # Matches your 0-20 grade scale
plt.grid(True, linestyle='--', alpha=0.3, axis='y') # Horizontal grid lines only
plt.tight_layout()

# Display the combined error bar plot
plt.show()
```
