```python
import numpy as np

# create a 1D array
array1 = np.array([1, 3, 5, 7, 9, 11])

# reshape the 1D array into a 2D array
array2 = np.reshape(array1, (2, 3))

# transpose the 2D array
array3 = np.transpose(array2)

print("Original array:\n", array1)
print("\nReshaped array:\n", array2)
print("\nTransposed array:\n", array3)
```

```python
# 1. Create an array of 5 random floating-point numbers between 0 and 1
float_array = np.random.rand(5)

# 2. Create an array of 5 random integers between 10 and 50
int_array = np.random.randint(10, 50, size=5)

print("Float Array:", float_array)
print("Integer Array:", int_array)
result = np.concatenate((float_array,int_array))
print(result)
unique_arr = np.unique(result)
print(unique_arr)
```

```python
original = np.array([1, 2, 3])
copied = original.copy()

# Modifying the copy does not change the original
copied[0] = 6
print(original)
print(copied)
arr=np.array([1, 2, 3, 4, 5])
v = arr.view()
# Modify an element in the view
v[0] = 10
# Both the original and the view show the change
print("Original:",arr)
print("View:", v)
```
