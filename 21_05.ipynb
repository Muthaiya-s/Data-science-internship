```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Generate the multi-column student dataset
np.random.seed(42)
names = ["Yadhu", "Ananya", "Rahul", "Sneha", "Vikram", "Priya"]
subjects = ["Maths", "Science", "English", "History", "Coding", "Art"]

data = []
for idx, name in enumerate(names, start=1001):
    for sub in subjects:
        mark = np.random.randint(35, 101)
        data.append({
            "Roll No": idx,
            "Student Name": name,
            "Subject": sub,
            "Marks": mark,
            "Status": "Pass" if mark >= 50 else "Fail"
        })
df = pd.DataFrame(data)

# 2. Initialize a 2x2 grid layout for comparison
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(2, 2, figsize=(16, 10), sharey=True)

# Graph 1: Box Plot (Distribution & Outliers)
sns.boxplot(data=df, x="Student Name", y="Marks", hue="Status",
            palette="Set2", ax=axes[0, 0])
axes[0, 0].set_title("1. Box Plot (Range & Spread Analysis)")

# Graph 2: Violin Plot (Density Shape)
sns.violinplot(data=df, x="Student Name", y="Marks", hue="Status",
               palette="Pastel1", ax=axes[0, 1], split=True, gap=0.1)
axes[0, 1].set_title("2. Violin Plot (Data Density Shape Analysis)")

# Graph 3: Bar Plot (Comparing Group Averages)
sns.barplot(data=df, x="Student Name", y="Marks", hue="Status",
            palette="muted", ax=axes[1, 0])
axes[1, 0].set_title("3. Bar Plot (Average Marks Comparison)")

# Graph 4: Swarm Plot (Every Individual Score Point)
sns.swarmplot(data=df, x="Student Name", y="Marks", hue="Status",
              palette="dark", ax=axes[1, 1], size=6, dodge=True)
axes[1, 1].set_title("4. Swarm Plot (Exact Point Visual Comparison)")

# 3. Clean up labels and legend placements across the dashboard
for ax in axes.flat:
    ax.set_xlabel("Student Name")
    ax.set_ylabel("Marks (Y-Axis)")
    sns.move_legend(ax, "lower left", title="Status")

plt.suptitle("Comparison Matrix: Student Performance Analysis", fontsize=16, fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Generate standard student data model
np.random.seed(42)
names = ["Yadhu", "Ananya", "Rahul", "Sneha", "Vikram", "Priya"]
subjects = ["Maths", "Science", "English", "History", "Coding", "Art"]

data = []
for idx, name in enumerate(names, start=1001):
    for sub in subjects:
        mark = np.random.randint(35, 101)
        data.append({
            "Roll No": idx,
            "Student Name": name,
            "Subject": sub,
            "Marks": mark,
            "Status": "Pass" if mark >= 50 else "Fail"
        })
df = pd.DataFrame(data)

# 2. Setup a 2x2 grid layout
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# Plot 1: Swarm Plot (Point Density Mapping)
sns.swarmplot(data=df, x="Student Name", y="Marks", hue="Status",
              palette="Dark2", dodge=True, size=6, ax=axes[0, 0])
axes[0, 0].set_title("1. Swarm Plot (Exact Point Density Visual)")

# Plot 2: Point Plot (Connecting Category Averages)
sns.pointplot(data=df, x="Student Name", y="Marks", hue="Status",
              palette="bright", errorbar=None, ax=axes[0, 1])
axes[0, 1].set_title("2. Point Plot (Average Trend Line Tracking)")

# Plot 3: Count Plot (Volume & Frequency Comparison)
sns.countplot(data=df, x="Student Name", hue="Status",
              palette="Set1", ax=axes[1, 0])
axes[1, 0].set_title("3. Count Plot (Total Pass vs Fail Quantities)")
axes[1, 0].set_ylabel("Number of Subjects")

# Plot 4: Strip Plot (Jittered Scatter Comparison)
sns.stripplot(data=df, x="Student Name", y="Marks", hue="Status",
              palette="deep", dodge=True, jitter=0.25, size=6, ax=axes[1, 1])
axes[1, 1].set_title("4. Strip Plot (Raw Values Distributed with Jitter)")

# 3. Clean up legends and axes labels globally
for ax in axes.flat:
    if ax != axes[1, 0]: # Keep subject count label intact on chart 3
        ax.set_ylabel("Marks Obtained")
    ax.set_xlabel("Student Name")
    sns.move_legend(ax, "lower left", title="Status")

plt.suptitle("Alternative Comparison Matrix: Advanced Seaborn Visualizations", fontsize=16, fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Generate an expanded dataset (Hours Studied vs Marks Obtained)
np.random.seed(42)
names = ["Yadhu", "Ananya", "Rahul", "Sneha", "Vikram", "Priya"]
subjects = ["Maths", "Science", "English", "History", "Coding", "Art"]

data = []
for idx, name in enumerate(names, start=1001):
    for sub in subjects:
        hours = np.random.randint(10, 50)
        # Marks are loosely tied to hours studied + some randomness
        mark = min(100, int(hours * 1.8 + np.random.randint(10, 30)))

        data.append({
            "Roll No": idx,
            "Student Name": name,
            "Subject": sub,
            "Hours Studied": hours,
            "Marks": mark,
            "Status": "Pass" if mark >= 50 else "Fail"
        })
df = pd.DataFrame(data)

# Set global Seaborn design theme
sns.set_theme(style="ticks")

# -------------------------------------------------------------
# GRAPH 1: Relational Regression Plot (Comparing Study Hours vs Marks)
# -------------------------------------------------------------
g1 = sns.lmplot(
    data=df,
    x="Hours Studied",
    y="Marks",
    hue="Status",
    palette={"Pass": "#2b7bba", "Fail": "#d95f02"},
    height=5,
    aspect=1.4
)
g1.set_titles("Study Time vs Marks Grade Comparison")


# -------------------------------------------------------------
# GRAPH 2: Multi-Grid Catplot (Isolated Subject Subplots)
# -------------------------------------------------------------
# col="Subject" dynamically spawns a separate subplot grid for each topic
g2 = sns.catplot(
    data=df,
    x="Student Name",
    y="Marks",
    hue="Status",
    col="Subject",
    col_wrap=3, # Wrap layout onto a new row after 3 columns
    kind="bar",
    palette="muted",
    height=3.5,
    aspect=1.2
)
g2.set_titles(col_template="{col_name} Performance Matrix")
g2.set_xticklabels(rotation=45)


# -------------------------------------------------------------
# GRAPH 3: Pair Plot Grid (Automated Column Relationships Matrix)
# -------------------------------------------------------------
# Extracts numerical comparisons grouped by Pass/Fail status
g3 = sns.pairplot(
    data=df[["Hours Studied", "Marks", "Status"]],
    hue="Status",
    palette="bright",
    corner=True # Cuts off duplicate mirror-image plots to keep it clean
)

plt.show()
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Generate the student dataset
np.random.seed(42)
names = ["Yadhu", "Ananya", "Rahul", "Sneha", "Vikram", "Priya"]
subjects = ["Maths", "Science", "English", "History", "Coding", "Art"]

data = []
for idx, name in enumerate(names, start=1001):
    for sub in subjects:
        mark = np.random.randint(35, 101)
        # Convert Status into a 1 or 0 binary flag for the KDE curve
        status_numeric = 1 if mark >= 50 else 0

        data.append({
            "Student Name": name,
            "Subject": sub,
            "Passed": status_numeric,
            "Status": "Pass" if mark >= 50 else "Fail"
        })
df = pd.DataFrame(data)

# 2. Set up the plotting theme
plt.figure(figsize=(10, 5))
sns.set_theme(style="whitegrid")

# 3. Create the Binary KDE Plot
# x="Passed" isolates the distribution strictly to 0 (Fail) and 1 (Pass)
# bw_adjust smoothens the curves across the two distinct categories
plot = sns.kdeplot(
    data=df,
    x="Passed",
    hue="Student Name",
    fill=True,
    common_norm=False,
    palette="muted",
    alpha=0.15,
    linewidth=2.5,
    bw_adjust=0.8
)

# 4. Swap the numeric 0 and 1 axis ticks out for clean "Fail" and "Pass" labels
plot.set_xticks([0, 1])
plot.set_xticklabels(["Fail (0)", "Pass (1)"])

# 5. Native Titles and Labels
plot.set_title("Student Pass vs Fail Density Comparison", fontsize=14, fontweight="bold", pad=15)
plot.set_xlabel("Academic Performance Outcome", fontsize=12)
plot.set_ylabel("Density Distribution", fontsize=12)
plot.set_xlim(-0.5, 1.5)

# Move Legend
sns.move_legend(plot, "upper center", title="Students", bbox_to_anchor=(0.5, -0.15), ncol=6)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# 1. Load the dataset (Make sure 'student-mat.csv' is in your Colab file directory)
df = pd.read_csv('student-mat.csv')

# 2. Convert Final Grades (G3) into a Binary 'Status' Column (Pass/Fail)
# Standard passing grade for this dataset is typically 10 out of 20
df['Status'] = df['G3'].apply(lambda x: 'Pass' if x >= 10 else 'Fail')

# 3. Convert Binary Categorical columns (yes/no) to numeric flags (1/0)
binary_cols = ['schoolsup', 'famsup', 'paid', 'activities', 'nursery', 'higher', 'internet', 'romantic']
# Adjusting column names slightly to match your specific dataframe's casing if needed:
# e.g., Interested_in_higher_edu, internet_access, romantic_relationship, etc.
rename_dict = {
    'extra_edu_supp': 'schoolsup',
    'family_edu_supp': 'famsup',
    'extra_paid_class': 'paid',
    'extra_curr_activities': 'activities',
    'Interested_in_higher_edu': 'higher',
    'internet_access': 'internet',
    'romantic_relationship': 'romantic'
}
df.rename(columns=rename_dict, inplace=True)

# Apply 1/0 conversion for standard yes/no columns
for col in ['schoolsup', 'famsup', 'paid', 'activities', 'nursery', 'higher', 'internet', 'romantic']:
    if col in df.columns:
        df[col] = df[col].map({'yes': 1, 'no': 0})

# 4. Convert structural categorical columns (like Mjob, Fjob) using One-Hot Encoding
df_converted = pd.get_dummies(df, columns=['Mother_job', 'Father_job', 'reason_to_chose_school', 'guardian'], drop_first=True)

# Preview your newly converted data frame
df_converted.head()
```

```python
# Create an average grade column across G1, G2, and G3
df_converted['Average_Grade'] = df_converted[['G1', 'G2', 'G3']].mean(axis=1)

# Preview the grade tracking columns
df_converted[['G1', 'G2', 'G3', 'Average_Grade', 'Status']].head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Melt the DataFrame to bring G1, G2, and G3 scores into a single column for comparison
df_melted = df_converted.melt(
    id_vars=['Status'],
    value_vars=['G1', 'G2', 'G3'],
    var_name='Grading Period',
    value_name='Grade'
)

# 2. Set the canvas layout
plt.figure(figsize=(9, 5))
sns.set_theme(style="whitegrid")

# 3. Create the box plot comparing G1, G2, and G3 split by Pass/Fail status
sns.boxplot(
    data=df_melted,
    x='Grading Period',
    y='Grade',
    hue='Status',
    palette={'Pass': '#2ecc71', 'Fail': '#e74c3c'}
)

# 4. Add clear formatting text labels
plt.title('Distribution Comparison of Grading Periods (G1 vs G2 vs G3)', fontsize=14, fontweight='bold')
plt.xlabel('Grading Period', fontsize=12)
plt.ylabel('Grades Obtained (0 - 20)', fontsize=12)
plt.ylim(-1, 21)
plt.legend(title='Final Status', loc='lower left')
plt.tight_layout()

# Display the plot
plt.show()
```
