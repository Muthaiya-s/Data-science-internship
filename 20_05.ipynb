```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Generate reproducible random marks between 40 and 100 for 6 students
np.random.seed(42)
students = ["Yadhu", "Ananya", "Rahul", "Sneha", "Vikram", "Priya"]
marks = np.random.randint(40, 101, size=6)

# Create a DataFrame
df = pd.DataFrame({
    "Student": students,
    "Marks": marks
})

# Sort by marks for a cleaner visual representation
df = df.sort_values(by="Marks", ascending=False).reset_index(drop=True)

# Plotting setup
plt.figure(figsize=(8, 5))
sns.set_theme(style="whitegrid")

# Create the barplot using a modern palette
ax = sns.barplot(x="Student", y="Marks", data=df, palette="viridis", hue="Student", legend=False)

# Add data labels on top of each bar
for p in ax.patches:
    ax.annotate(f"{int(p.get_height())}",
                (p.get_x() + p.get_width() / 2., p.get_height() - 6),
                ha='center', va='center',
                color='white', fontweight='bold', fontsize=11)

plt.title("Student Performance Analysis (Dummy Data)", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Students", fontsize=12)
plt.ylabel("Marks Obtained", fontsize=12)
plt.ylim(0, 110) # Give extra space at the top for labels

# Save figure to display
plt.savefig("student_marks_analysis.png", bbox_inches='tight', dpi=150)
plt.close()

# Print summary data to display in text
print(df.to_string(index=False))
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Generate random dummy marks (between 40 and 100) for 6 students
np.random.seed(42)  # Keeps the random scores consistent every time you run it
students = ["Yadhu", "Ananya", "Rahul", "Sneha", "Vikram", "Priya"]
marks = np.random.randint(40, 101, size=6)

# 2. Structure the data into a DataFrame and sort it (highest score to lowest)
df = pd.DataFrame({
    "Student": students,
    "Marks": marks
}).sort_values(by="Marks", ascending=False).reset_index(drop=True)

# 3. Set the layout and visual background theme
plt.figure(figsize=(8, 5))
sns.set_theme(style="whitegrid")

# 4. Draw the bar plot using Seaborn
# (Using a solid deep blue color '#345c85' to keep the bars clean and uniform)
ax = sns.barplot(x="Student", y="Marks", data=df, color="#345c85")

# 5. Loop through each bar to draw the numeric score value inside it
for p in ax.patches:
    h = p.get_height()
    if not np.isnan(h):
        ax.annotate(
            f"{int(h)}",
            (p.get_x() + p.get_width() / 2., h - 5), # Sets text position inside top of the bar
            ha='center',
            va='center',
            color='white',
            fontweight='bold',
            fontsize=11
        )

# 6. Customize titles, axis labels, and vertical limits
plt.title("Student Performance Analysis", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Students", fontsize=12)
plt.ylabel("Marks Obtained", fontsize=12)
plt.ylim(0, 110) # Gives a tiny bit of breathing room above the tallest bar

# 7. Render the final plot
plt.show()
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Setup the 6 students and 6 subjects
students = ["Yadhu", "Ridhu", "Elakiya", "Arun", "Balaji", "Priya"]
subjects = ["Maths", "Science", "English", "History", "Coding", "Art"]

# 2. Generate random marks (between 45 and 100) for each student across all subjects
np.random.seed(42)
data_list = []

for student in students:
    for subject in subjects:
        random_mark = np.random.randint(45, 101)
        data_list.append({"Student": student, "Subject": subject, "Marks": random_mark})

# Convert to a "Long-Form" DataFrame (ideal for Seaborn)
df = pd.DataFrame(data_list)

# 3. Setup the plotting window size
plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid")

# 4. Create a Grouped Bar Plot
# x="Student" groups them on the axis
# hue="Subject" splits each student's section into 6 individual subject bars
ax = sns.barplot(x="Student", y="Marks", hue="Subject", data=df, palette="muted")

# 5. Fine-tune the chart details
plt.title("Student Multi-Subject Performance Analysis", fontsize=15, fontweight='bold', pad=15)
plt.xlabel("Students", fontsize=12, fontweight='bold')
plt.ylabel("Marks Obtained", fontsize=12, fontweight='bold')
plt.ylim(0, 115)  # Make room for the legend

# Move the subject legend outside or to a clean corner so it doesn't block the bars
plt.legend(title="Subjects", bbox_to_anchor=(1.02, 1), loc='upper left', borderaxespad=0)

# Clean up layout and render
plt.tight_layout()
plt.show()
```
